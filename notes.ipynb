7-Read the total profit of all months and show it using the Bar plot

```python
df.plot(x='month_number', y='total_profit', kind='bar', color='tab:blue', legend = False)

# Set required properties

plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
plt.grid(alpha=0.5, linestyle='--')
# Display the plot
plt.show()
```

8-Read the total profit of all of the months, and show the line plot with the following style properties:
The generated line plot must include the following Style properties:

Show the legend at the lower-right location.
X label name = Month number
Y label name = Total profit.
Title = Company profit per month.
Add a circle marker; the line marker color = black.

```python

monthList = df['month_number'].tolist()
profitList = df['total_profit'].tolist()

# 2. Create the line plot with the requested styling properties
plt.plot(
    monthList, 
    profitList, 
    label='Profit data of last year', 
    color='red',             
    linestyle=':',           
    linewidth=3,             
    marker='o',              
    markerfacecolor='black', 
    markeredgecolor='black'
)


plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
plt.legend(loc='lower right') 

plt.xticks(monthList)


plt.show()
```

9-Print all of the product sales data and show it using a multi-line plot.
Display the number of units sold per month for each product using multi-line plots. (i.e., show a separate plotline for each product).


```python
monthList = df['month_number'].tolist()
faceCreamSalesData = df['facecream'].tolist()
faceWashSalesData = df['facewash'].tolist()
toothPasteSalesData = df['toothpaste'].tolist()
bathingSoapSalesData = df['bathingsoap'].tolist()
shampooSalesData = df['shampoo'].tolist()
moisturizerSalesData = df['moisturizer'].tolist()

# 2. Create multi-line plots for each product
plt.plot(monthList, faceCreamSalesData, label='Face cream sales data', marker='o', linewidth=3)
plt.plot(monthList, faceWashSalesData, label='Face wash sales data', marker='o', linewidth=3)
plt.plot(monthList, toothPasteSalesData, label='Toothpaste sales data', marker='o', linewidth=3)
plt.plot(monthList, bathingSoapSalesData, label='Bathing soap sales data', marker='o', linewidth=3)
plt.plot(monthList, shampooSalesData, label='Shampoo sales data', marker='o', linewidth=3)
plt.plot(monthList, moisturizerSalesData, label='Moisturizer sales data', marker='o', linewidth=3)

# 3. Configure labels, title, and legend
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.title('Sales data')
plt.xticks(monthList)
plt.legend(loc='upper left')

# 4. Display the plot
plt.show()

```

A line chart of date versus temperature must include the following properties:

```python


date = ["25/12", "26/12", "27/12"]
temp = [8.5, 10.5, 6.8]

```

```python

plt.figure(figsize=(6, 4))
plt.plot(date, temp, marker='o', linestyle='-', color='b')

plt.xlabel("Date")
plt.ylabel("temperature")
plt.title("Date-wise Temperature")

plt.grid(True)
plt.tight_layout()
plt.show()
```

weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]

Show the average weight against the average height (use a line chart or any other chart):
The generated line plot must include the following Style properties:
The line style should be dash-dot and the line color should be green.
Show the legend at the lower-right location.
X label name = Weight in kg.
Y label name = Height in cm.
Title = Average weight with respect to average height.
Add a circle marker: marker size = 10; circle marker color = green.


```python
height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]
```

```python
plt.figure(figsize=(7, 5))
plt.plot(weight, height, 
         linestyle='-.', 
         color='green', 
         marker='o', 
         markersize=10, 
         markerfacecolor='green', 
         label='Height vs Weight')


plt.xlabel("Weight in kg.")
plt.ylabel("Height in cm.")
plt.title("Average weight with respect to average height")



plt.tight_layout()
plt.show()
```
